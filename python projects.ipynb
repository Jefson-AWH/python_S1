```python
#program to accept principal amount, no of years,rate of interest and find simple interset
principal=float(input("enter the principal amount:"))
years=float(input("enter the  no of years: "))
rate=float(input("enter the rate of interest:"))
simple_interest=(principal*years*rate)/100
print("the simple interest rate is:",simple_interest)
```
